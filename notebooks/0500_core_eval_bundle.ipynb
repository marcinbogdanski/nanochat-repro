# Core Eval

Get eval bundle from NanoChat, unzip and put in `data` folder

```python
EVAL_BUNDLE = "https://karpathy-public.s3.us-west-2.amazonaws.com/eval_bundle.zip"

# Download and unzip the eval bundle if it doesn't exist
import os
import urllib.request
import zipfile

os.makedirs('../data', exist_ok=True)

filepath = '../data/eval_bundle.zip'

if not os.path.exists(filepath):
    print("Downloading eval bundle...")
    urllib.request.urlretrieve(EVAL_BUNDLE, '../data/eval_bundle.zip')

print("Unzipping eval bundle...")
with zipfile.ZipFile('../data/eval_bundle.zip', 'r') as zip_ref:
    zip_ref.extractall('../data')
print("Eval bundle is ready.")
```
