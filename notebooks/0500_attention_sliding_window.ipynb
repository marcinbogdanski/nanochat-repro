# Attention Mask for SDPA

Short notebook to play with how to make masks for SDPA

```python
import torch
import torch.nn.functional as F
```

```python
Tq = 5
Tk = 8
```

```python
row_idx = (Tk-Tq) + torch.arange(Tq)
row_idx = row_idx.unsqueeze(1)
row_idx
```

```python
col_idx = torch.arange(Tk).unsqueeze(0)
col_idx
```

```python
attention_mask = col_idx <= row_idx
attention_mask.to(torch.int64)
```

```python
window = 2

in_window_mask = (row_idx - col_idx) <= window
in_window_mask.to(torch.int64)
```

```python
final_mask = attention_mask & in_window_mask
final_mask.to(torch.int64)
```
